```python
import pandas as pd
import numpy as np
```

# Inspecter les colonnes et types de données

```python
df= pd.read_csv("store.csv")
print(df.head(10))
print(df.info())
print(df.describe())
```

# Analyser la distribution des valeurs et repérer les anomalies

```python
Q1 = df["Postal Code"].quantile(0.25)
Q3 = df["Postal Code"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 - 1.5 * IQR

outeliers = df[(df["Postal Code"] < lower_bound) | (df["Postal Code"] > upper_bound)]
print(outeliers["Postal Code"].head())
```

# Identifier et gérer les valeurs manquantes

```python
print(df.isnull().sum())
```

# Supprimer ou corriger les doublons

```python
df = df.drop_duplicates()
print(df)
```

```python
print(df.dtypes)
print(df)
df.rename(columns={"Sales;": "Sales"}, inplace=True)
print(df)
```

# Corriger les types de colonnes (dates, numériques, catégories)

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], errors='coerce', dayfirst= True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors='coerce', dayfirst= True)
print(df.dtypes)
```

```python
df["Postal Code"] = df["Postal Code"].fillna(0).astype(int)
df["Sales"] = df["Sales"].astype(str)
df["Sales"] = df["Sales"].str.replace(",", "").str.replace("$", "").str.strip()
df["Sales"] = pd.to_numeric(df["Sales"], errors='coerce')
print(df.dtypes)
```

```python
col_cat = ["Ship Mode","Segment","Country","City","State","Region","Category","Sub-Category"]
for col in col_cat:
    df[col] = df[col].astype("category")
print(df.dtypes)
```

# Extraire des informations temporelles : année, mois, trimestre

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], errors='coerce', dayfirst= True)
df["year"] = df["Order Date"].dt.year
df["month"] = df["Order Date"].dt.month
df["quarter"] = df["Order Date"].dt.quarter
print(df[["Order Date","year", "month", "quarter"]].head())
```

# Créer de nouvelles métriques : marge, taux de profit, ratio quantité / ventes

```python
df["cout"] = df["Sales"]*0.8
df["profit"] = df["Sales"] - df["cout"]
df["taux_profit"] = df["profit"]/df["Sales"]*100
df["ratio"] = df["profit"]/df["Sales"]
print(df)
print(df[df["Sales"].notna()])
```

```python
from utilis import segmentation

df["segmentation"] = df["Sales"].map(segmentation)
print(df)
print(df[df["Sales"].notna()])

```

# délais livraison

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)
df["delai_laivraison"] = (df["Ship Date"] - df["Order Date"]).dt.days
print(df)
```

#  ventes par produit ou région

```python
vente_par_produit = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False)
print(vente_par_produit)
```

```python
vente_par_region = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
print(vente_par_region)
```

```python
vente_totale = df.groupby("Region")["Sales"].agg(["sum", "mean", "max", "min"])
print(vente_totale)
```
